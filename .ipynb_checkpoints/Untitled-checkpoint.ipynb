````python
from pathlib import Path
import json
import h5py
import pandas as pd
from pynwb import NWBHDF5IO
from pynwb.misc import DynamicTable

repo = Path("/Users/euo9382/Documents/Repositories/analysis_Belal2026")

counts_csv = repo / "RNAscope data" / "RAW" / "RNAscope_experimenter.csv"
nwb_root = repo / "RNAscope data" / "NWB"

counts_df = pd.read_csv(counts_csv)

rnascope_meta = {
    "dataset_name": "Figure 11 RNAscope local NWB conversion",
    "publication_status": "local working metadata; final archive identifier/version not assigned here",
    "description": (
        "Multiplex fluorescent RNAscope dataset examining Chrnb2 transcript expression in dorsal "
        "striatal interneurons from control and unilateral 6-OHDA lesioned mice. Tissue was collected "
        "four weeks after lesioning and labeled with probes targeting CHRNB2, TH, and NDNF to identify "
        "TH-expressing interneurons and NDNF-expressing neurogliaform interneurons. Sections were imaged "
        "with confocal microscopy, and Chrnb2-positive puncta were quantified within manually defined "
        "cellular ROIs. The dataset compares Chrnb2 signal per cell between intact and lesioned sides "
        "of the striatum."
    ),
    "contributors": [
        {
            "name": "Zhong, Xie",
            "email": "z-xie@northwestern.edu",
            "role": ["dcite:Author", "dcite:DataCollector"],
            "affiliation": "Northwestern University",
        },
        {
            "name": "Surmeier, D. James",
            "email": "j-surmeier@northwestern.edu",
            "role": ["dcite:Author", "dcite:ContactPerson"],
            "affiliation": "Northwestern University",
        },
    ],
    "license": ["spdx:CC-BY-4.0"],
    "keywords": [
        "RNAscope",
        "multiplex fluorescent in situ hybridization",
        "confocal microscopy",
        "mouse",
        "striatum",
        "dorsal striatum",
        "TH interneuron",
        "neurogliaform interneuron",
        "NDNF",
        "CHRNB2",
        "nicotinic acetylcholine receptor",
        "6-OHDA",
        "dopamine depletion",
        "Parkinson disease model",
    ],
    "protocol": (
        "ChAT-Cre mice were sacrificed four weeks after unilateral 6-OHDA lesioning. Brain tissue was "
        "rapidly collected, embedded in OCT, flash frozen, and sectioned coronally at 12 um. RNAscope "
        "multiplex fluorescent in situ hybridization was performed on fresh frozen sections using the "
        "RNAscope Multiplex Fluorescent Detection Kit v2 with probes against CHRNB2, NDNF, and TH. "
        "Slides were mounted with DAPI-containing antifade medium and imaged on an Olympus FluoView FV10i "
        "confocal microscope. TH-positive and NDNF-positive cells in dorsal striatum were manually "
        "outlined as ROIs, and Chrnb2-positive puncta within each ROI were counted."
    ),
    "relatedResource": [
        {
            "identifier": "10.17504/protocols.io.14egn3odml5d/v1",
            "name": "v2 RNAscope",
            "relation": "dcite:IsDescribedBy",
            "url": "https://dx.doi.org/10.17504/protocols.io.14egn3odml5d/v1",
            "resourceType": "dcite:Method",
        }
    ],
}

count_methodology = {
    "name": "Experimenter Chrnb2 puncta counts",
    "description": (
        "Chrnb2-positive puncta were counted within manually defined cellular ROIs. ROIs correspond "
        "to TH-positive or NDNF-positive interneurons. Counts are stored per ROI/replicate and linked "
        "to the RNAscope imaging field, hemisphere, condition, session, and cell type."
    ),
    "source_file": str(counts_csv),
    "count_column": "count",
    "roi_identifier_column": "replicate",
}

sessions = ["L1.ST8", "L2.ST8", "L3.ST6", "L4.ST8"]

for session in sessions:
    nwb_path = nwb_root / session / f"{session}.nwb"
    session_df = counts_df[counts_df["session"].astype(str) == session].copy()

    if not nwb_path.exists():
        print(f"Missing NWB, skipping: {nwb_path}")
        continue

    if session_df.empty:
        print(f"No experimenter counts for {session}, skipping")
        continue

    module_name = "rnascope_analysis_metadata"
    table_name = "experimenter_chrnb2_counts"

    custom_meta = {
        "Figure 11 RNAscope": {
            "local_conversion_metadata": rnascope_meta,
            "count_methodology": count_methodology,
            "session": session,
        }
    }

    custom_json = json.dumps(custom_meta, indent=2)

    # Delete old objects at the HDF5 level before PyNWB reads the file.
    # This makes reruns deterministic.
    with h5py.File(nwb_path, "r+") as f:
        old_paths = [
            f"processing/{module_name}/{table_name}",
            "scratch/session_metadata_custom",
        ]
        for old_path in old_paths:
            if old_path in f:
                del f[old_path]

    with NWBHDF5IO(str(nwb_path), "r+", load_namespaces=True) as io:
        nwbfile = io.read()

        # Top-level NWB metadata.
        nwbfile.fields["session_description"] = f"{session} RNAscope imaging session"
        nwbfile.fields["experimenter"] = ["Zhong, Xie", "Surmeier, D. James"]
        nwbfile.fields["lab"] = "Surmeier Lab"
        nwbfile.fields["institution"] = "Northwestern University"
        nwbfile.fields["experiment_description"] = rnascope_meta["description"]
        nwbfile.fields["protocol"] = rnascope_meta["protocol"]
        nwbfile.fields["keywords"] = rnascope_meta["keywords"]

        # Full contributor/contact/method metadata.
        nwbfile.add_scratch(
            custom_json,
            name="session_metadata_custom",
            description="Local RNAscope metadata, contributor metadata, and count methodology",
        )

        # Check counts fields exist in this NWB.
        field_table = (
            nwbfile.processing["rnascope_source_metadata"]["field_metadata"]
            .to_dataframe()
            .reset_index(drop=True)
        )
        nwb_fields = set(field_table["field_name"].astype(str))
        csv_fields = set(session_df["field"].astype(str))

        missing_fields = sorted(csv_fields - nwb_fields)
        if missing_fields:
            raise ValueError(f"{session}: count fields missing from NWB: {missing_fields}")

        # Add experimenter counts table.
        mod = nwbfile.processing.get(module_name)
        if mod is None:
            mod = nwbfile.create_processing_module(
                name=module_name,
                description="Post-conversion RNAscope analysis metadata and experimenter counts",
            )

        table = DynamicTable(
            name=table_name,
            description="Experimenter Chrnb2 puncta counts imported from RNAscope_experimenter.csv",
        )

        table.add_column("condition", "Condition label: Intact or Lesioned")
        table.add_column("cell_type", "ROI/cell type label: NDNF+ or TH+")
        table.add_column("slice_id", "Original slice/session token")
        table.add_column("field", "RNAscope imaging field name")
        table.add_column("hemisphere", "Hemisphere token: UL or L")
        table.add_column("field_index", "Field index within hemisphere/session")
        table.add_column("replicate", "ROI/replicate index")
        table.add_column("count", "Experimenter Chrnb2 puncta count")
        table.add_column("session", "Session folder name")
        table.add_column("session_group", "Session group label")

        for _, row in session_df.iterrows():
            table.add_row(
                condition=str(row["condition"]),
                cell_type=str(row["cell_type"]),
                slice_id=str(row["slice_id"]),
                field=str(row["field"]),
                hemisphere=str(row["hemisphere"]),
                field_index=int(row["field_index"]),
                replicate=int(row["replicate"]),
                count=int(row["count"]),
                session=str(row["session"]),
                session_group=str(row["session_group"]),
            )

        mod.add(table)
        nwbfile.set_modified()
        io.write(nwbfile)

    print(f"Updated {nwb_path} with {len(session_df)} experimenter count rows")
````

````python
from pathlib import Path
import json
import pandas as pd
from pynwb import NWBHDF5IO
from IPython.display import display, Markdown

repo = Path("/Users/euo9382/Documents/Repositories/analysis_Belal2026")
nwb_root = repo / "RNAscope data" / "NWB"

nwb_path = nwb_root / "L1.ST8" / "L1.ST8.nwb"

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
pd.set_option("display.max_colwidth", None)

with NWBHDF5IO(str(nwb_path), "r", load_namespaces=True) as io:
    nwbfile = io.read()

    nwb_meta = {
        "session_description": nwbfile.session_description,
        "identifier": nwbfile.identifier,
        "session_start_time": nwbfile.session_start_time,
        "experiment_description": nwbfile.experiment_description,
        "experimenter": list(nwbfile.experimenter) if nwbfile.experimenter is not None else None,
        "lab": nwbfile.lab,
        "institution": nwbfile.institution,
        "protocol": nwbfile.protocol,
        "notes": nwbfile.notes,
        "keywords": list(nwbfile.keywords) if nwbfile.keywords is not None else None,
    }

    subject_meta = {}
    if nwbfile.subject is not None:
        subject_meta = {
            "subject_id": nwbfile.subject.subject_id,
            "description": nwbfile.subject.description,
            "species": nwbfile.subject.species,
            "sex": nwbfile.subject.sex,
            "genotype": nwbfile.subject.genotype,
            "age": nwbfile.subject.age,
            "strain": nwbfile.subject.strain,
        }

    counts = (
        nwbfile.processing["rnascope_analysis_metadata"]["experimenter_chrnb2_counts"]
        .to_dataframe()
        .reset_index(drop=True)
    )

    custom_raw = nwbfile.scratch["session_metadata_custom"].data
    if isinstance(custom_raw, bytes):
        custom_raw = custom_raw.decode()
    custom_meta = json.loads(custom_raw)

display(Markdown(f"# NWB Metadata: `{nwb_path.name}`"))

display(Markdown("## NWBFile"))
display(pd.DataFrame([nwb_meta]).T.rename(columns={0: "value"}))

display(Markdown("## Subject"))
display(pd.DataFrame([subject_meta]).T.rename(columns={0: "value"}))

display(Markdown("## Custom Metadata"))
display(Markdown(f"```json\n{json.dumps(custom_meta, indent=2)}\n```"))

display(Markdown("## Full Experimenter Counts"))
display(
    counts.sort_values(
        ["condition", "cell_type", "session", "hemisphere", "field_index", "replicate"],
        kind="stable",
    ).reset_index(drop=True)
)
````
